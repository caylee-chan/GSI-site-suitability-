```python
# Libraries
import xarray as xr # For viewing NetCDF files
import rasterio 
import numpy as np
import rioxarray
```

## Cx. pipiens

```python
# Culex pipiens

# Load in data
pipiens = './data/mosquitoes/culex_pipiens_meansuitability.nc'
xrds = xr.open_dataset(pipiens, engine = 'netcdf4') 
print(f"Shape: {xrds.data_vars['layer'].shape}")
xrds
```

```python
# Culex pipiens
xrds['layer']
```

```python
# Export as geotiff - https://gis.stackexchange.com/questions/323317/converting-netcdf-dataset-array-to-geotiff-using-rasterio-python
out_fp = './data/mosquitoes/culex_pipiens_meansuitability.tif'

# Add CRS
xrds.rio.write_crs('epsg:4326', inplace = True)

# Write
xrds['layer'].rio.to_raster(out_fp)
```

```python
# Close file
xrds.close()
```
