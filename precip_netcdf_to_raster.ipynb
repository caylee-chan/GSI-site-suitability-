```python
# Libraries
import xarray as xr # For viewing NetCDF files
import rasterio 
import numpy as np
import rioxarray
```

```python
#### ssp2 2025: ############################################################################################################
```

```python
# Load in data
ssp2_2025 = './data/projected_precip/precip_ssp2_2025/GISS-E2-1-G_ssp2_2025_pr.nc'
xrds = xr.open_dataset(ssp2_2025, engine = 'netcdf4') 
print(f"Shape: {xrds.data_vars['pr'].shape}")
xrds
```

```python
# Prcp var
xrds['pr']
```

```python
# Convert units to mm/day - https://stackoverflow.com/questions/75662038/convert-units-in-a-netcdf-file-efficiently
pr_mm_day = xrds['pr'] * 86400
pr_mm_day
```

```python
# Calculate mean prcp rate (for all days)
pr_2025_mean_prcp_rate = pr_mm_day.mean(dim = 'time')
print(f'Shape: {pr_2025_mean_prcp_rate.shape}')
pr_2025_mean_prcp_rate
```

```python
# Calculate sd of prcp rate
pr_2025_sd_prcp_rate = pr_mm_day.std(dim = 'time')
print(f'Shape: {pr_2025_sd_prcp_rate.shape}')
pr_2025_sd_prcp_rate
```

```python
# Export mean prcp rate as geotiff
out_mean_prcp_ssp2_2025 = './data/projected_precip/precip_ssp2_2025/ssp2_2025_prcp_rate.tif'

# Add CRS
pr_2025_mean_prcp_rate.rio.write_crs('epsg:4326', inplace = True)

# Write
pr_2025_mean_prcp_rate.rio.to_raster(out_mean_prcp_ssp2_2025)


# Export mean sd as geotiff
out_sd_prcp_ssp2_2025 = './data/projected_precip/precip_ssp2_2025/ssp2_2025_standard_dev_prcp_rate.tif'

# Add CRS
pr_2025_sd_prcp_rate.rio.write_crs('epsg:4326', inplace = True)

# Write
pr_2025_sd_prcp_rate.rio.to_raster(out_sd_prcp_ssp2_2025)
```

```python
# Close file
xrds.close()
```
